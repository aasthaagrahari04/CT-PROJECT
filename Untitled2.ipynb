```python
import g4f

def get_word_details(word):
    try:
        # 🔥 Few-shot prompting
        messages = [
            {"role": "system", "content": "You are a helpful English dictionary."},

            {"role": "user", "content": "Word: ambition"},
            {"role": "assistant", "content": """Word: ambition
Meaning: A strong desire to achieve something
Example: Her ambition is to become a doctor
Synonyms: goal, aspiration"""},

            {"role": "user", "content": "Word: resilience"},
            {"role": "assistant", "content": """Word: resilience
Meaning: Ability to recover quickly from difficulties
Example: His resilience helped him overcome failure
Synonyms: strength, toughness"""},

            {"role": "user", "content": f"Word: {word}"}
        ]

        response = g4f.ChatCompletion.create(
            model="gpt-4",
            messages=messages
        )

        return response

    except Exception as e:
        return f"❌ Error: {e}"
```

```python
word = input("Enter a word (or press Enter for demo): ")

if word.strip() == "":
    print("\n👉 Running Example Words...\n")
    demo_words = ["focus", "confidence", "discipline"]

    for w in demo_words:
        print(get_word_details(w))
        print("-" * 50)
else:
    print("\n📘 Word Details:\n")
    print(get_word_details(word))
```
